```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.lines import Line2D
from matplotlib.patches import Circle, Wedge
```

```python
G = nx.Graph()

# Stations (Nodes)
stations = [
    "Oxford Circus",
    "Piccadilly Circus",
    "Leicester Square",
    "Covent Garden",
    "Charing Cross"
]

G.add_nodes_from(stations)

# Railway Connections (Edges)
# Format:
# (Station 1, Station 2, Distance in km)

edges = [
    ("Oxford Circus", "Piccadilly Circus", 0.8),
    ("Piccadilly Circus", "Charing Cross", 1.1),
    ("Piccadilly Circus", "Leicester Square", 0.6),
    ("Leicester Square", "Charing Cross", 0.7),
    ("Leicester Square", "Covent Garden", 0.5)
]

for station1, station2, distance in edges:
    G.add_edge(station1, station2, weight=distance)

# Position of each station
pos = {
    "Oxford Circus": (0, 4),
    "Piccadilly Circus": (1, 2),
    "Leicester Square": (2, 2),
    "Covent Garden": (3, 3),
    "Charing Cross": (2, 0)
}

# Create Figure
plt.figure(figsize=(10, 7))

# Draw Stations
nx.draw_networkx_nodes(
    G,
    pos,
    node_color="lightgray",
    node_size=450
)

# Underground Lines
lines = {
    "Bakerloo": {
        "color": "#996633",
        "edges": [
            ("Oxford Circus", "Piccadilly Circus"),
            ("Piccadilly Circus", "Charing Cross")
        ]
    },

    "Piccadilly": {
        "color": "royalblue",
        "edges": [
            ("Piccadilly Circus", "Leicester Square"),
            ("Leicester Square", "Covent Garden")
        ]
    },

    "Northern": {
        "color": "black",
        "edges": [
            ("Leicester Square", "Charing Cross")
        ]
    }
}

# Draw each underground line
for line_name, line in lines.items():

    nx.draw_networkx_edges(
        G,
        pos,
        edgelist=line["edges"],
        edge_color=line["color"],
        width=3
    )

edge_labels = {
    (station1, station2): f"{distance} km"
    for station1, station2, distance in edges
}

nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=edge_labels,
    font_size=9
)

label_pos = {
    "Oxford Circus": (0.10, 3.95),
    "Piccadilly Circus": (0.40, 1.90),
    "Leicester Square": (2.10, 1.80),
    "Covent Garden": (3.10, 2.95),
    "Charing Cross": (1.49, -0.10)
}

for station, (x, y) in label_pos.items():
    plt.text(
        x,
        y,
        station,
        fontsize=10
    )

for line_name, line in lines.items():

    plt.plot(
        [],
        [],
        color=line["color"],
        linewidth=3,
        label=line_name
    )

plt.legend(
    title="Key",
    loc="lower right",
    frameon=True,
    fontsize=12,
    title_fontsize=13
)

plt.title(
    "London Underground Network",
    fontsize=15,
    weight="bold"
)

plt.axis("off")

plt.show()
```

```python
# Create Graph

G = nx.Graph()

# Stations
# Total: 21 stations

stations = [
    # Bakerloo-related stations
    "Oxford Circus",
    "Piccadilly Circus",
    "Charing Cross",
    "Embankment",
    "Waterloo",
    "Lambeth North",

    # Piccadilly-related stations
    "South Kensington",
    "Green Park",
    "Leicester Square",
    "Covent Garden",
    "Holborn",

    # Northern-related stations
    "Camden Town",
    "Tottenham Court Road",

    # Central-related stations
    "Marble Arch",
    "Bond Street",
    "Bank",

    # Jubilee-related stations
    "Westminster",
    "Southwark",
    "London Bridge",
    "Bermondsey",
    "Canada Water"
]

G.add_nodes_from(stations)


# Railway Connections
# Format: (Station 1, Station 2, Distance)

edges = [

    # Bakerloo Line

    ("Oxford Circus", "Piccadilly Circus", 0.9),
    ("Piccadilly Circus", "Charing Cross", 0.7),
    ("Charing Cross", "Embankment", 0.4),
    ("Embankment", "Waterloo", 0.7),
    ("Waterloo", "Lambeth North", 0.6),


    # Piccadilly Line

    ("South Kensington", "Green Park", 2.73),
    ("Green Park", "Piccadilly Circus", 0.68),
    ("Piccadilly Circus", "Leicester Square", 0.5),
    ("Leicester Square", "Covent Garden", 0.3),
    ("Covent Garden", "Holborn", 0.6),


    # Northern Line

    ("Camden Town", "Tottenham Court Road", 3.0),
    ("Tottenham Court Road", "Leicester Square", 1.0),
    ("Leicester Square", "Charing Cross", 0.4),
    ("Charing Cross", "Embankment", 0.3),
    ("Embankment", "Waterloo", 0.7),


    # Central Line

    ("Marble Arch", "Bond Street", 0.6),
    ("Bond Street", "Oxford Circus", 0.6),
    ("Oxford Circus", "Tottenham Court Road", 0.8),
    ("Tottenham Court Road", "Holborn", 0.8),
    ("Holborn", "Bank", 2.3),


    # Jubilee Line

    ("Westminster", "Waterloo", 1.0),
    ("Waterloo", "Southwark", 0.5),
    ("Southwark", "London Bridge", 1.2),
    ("London Bridge", "Bermondsey", 2.1),
    ("Bermondsey", "Canada Water", 1.1)
]


# Add connections to graph
for station1, station2, distance in edges:
    G.add_edge(
        station1,
        station2,
        weight=distance
    )


# Position of Stations

pos = {
    # Bakerloo corridor
    "Oxford Circus": (2, 9),
    "Piccadilly Circus": (3, 7),
    "Charing Cross": (4, 5),
    "Embankment": (5, 4),
    "Waterloo": (6, 2),
    "Lambeth North": (6, 0),

    # Piccadilly corridor
    "South Kensington": (0, 5),
    "Green Park": (1, 7),
    "Leicester Square": (5, 7),
    "Covent Garden": (6.45, 8),
    "Holborn": (8, 9),

    # Northern corridor
    "Camden Town": (5, 11),
    "Tottenham Court Road": (5, 9.45),

    # Central corridor
    "Marble Arch": (0, 9),
    "Bond Street": (1, 10),
    "Bank": (11, 8.75),

    # Jubilee corridor
    "Westminster": (4, 2.25),
    "Southwark": (8, 1.60),
    "London Bridge": (10, 2.65),
    "Bermondsey": (12, 2),
    "Canada Water": (14, 1.80)
}


# Label Positions

label_pos = {
    station: (x, y + 0.38)
    for station, (x, y) in pos.items()
}

label_pos.update({
    "South Kensington": (0.16, 4.50),
    "Green Park": (1.00, 7.40),
    "Oxford Circus": (2.50, 9.35),
    "Tottenham Court Road": (6.25, 9.65),
    "Camden Town": (5.00, 11.42),

    "Piccadilly Circus": (3.80, 7.35),
    "Leicester Square": (6.05, 6.95),
    "Covent Garden": (7.25, 7.75),
    "Holborn": (8.00, 8.55),
    "Bank": (11.00, 8.42),

    "Charing Cross": (3.15, 5.10),
    "Embankment": (4.15, 3.95),

    "Westminster": (4.00, 1.90),
    "Waterloo": (6.45, 2.30),
    "Southwark": (8.00, 1.30),
    "London Bridge": (10.00, 2.95),
    "Bermondsey": (12.00, 1.70),
    "Canada Water": (14.00, 2.15),

    "Lambeth North": (6.75, 0.00),
    "Marble Arch":(0, 8.65),
    "Camden Town": (5, 11.3)
})


# Underground Lines
# Total: 5 lines

lines = {
    "Bakerloo": {
        "color": "#996633",
        "shape": "s",
        "edges": [
            ("Oxford Circus", "Piccadilly Circus"),
            ("Piccadilly Circus", "Charing Cross"),
            ("Charing Cross", "Embankment"),
            ("Embankment", "Waterloo"),
            ("Waterloo", "Lambeth North")
        ]
    },

    "Piccadilly": {
        "color": "#1F4EBA",
        "shape": "D",
        "edges": [
            ("South Kensington", "Green Park"),
            ("Green Park", "Piccadilly Circus"),
            ("Piccadilly Circus", "Leicester Square"),
            ("Leicester Square", "Covent Garden"),
            ("Covent Garden", "Holborn")
        ]
    },

    "Northern": {
        "color": "#111111",
        "shape": "^",
        "edges": [
            ("Camden Town", "Tottenham Court Road"),
            ("Tottenham Court Road", "Leicester Square"),
            ("Leicester Square", "Charing Cross"),
            ("Charing Cross", "Embankment"),
            ("Embankment", "Waterloo")
        ]
    },

    "Central": {
        "color": "#E32017",
        "shape": "h",
        "edges": [
            ("Marble Arch", "Bond Street"),
            ("Bond Street", "Oxford Circus"),
            ("Oxford Circus", "Tottenham Court Road"),
            ("Tottenham Court Road", "Holborn"),
            ("Holborn", "Bank")
        ]
    },

    "Jubilee": {
        "color": "#34A99D",
        "shape": "p",
        "edges": [
            ("Westminster", "Waterloo"),
            ("Waterloo", "Southwark"),
            ("Southwark", "London Bridge"),
            ("London Bridge", "Bermondsey"),
            ("Bermondsey", "Canada Water")
        ]
    }
}


# Create Station-to-Line Mapping

station_to_lines = {
    station: []
    for station in stations
}

for line_name, line_information in lines.items():
    for station1, station2 in line_information["edges"]:

        if line_name not in station_to_lines[station1]:
            station_to_lines[station1].append(line_name)

        if line_name not in station_to_lines[station2]:
            station_to_lines[station2].append(line_name)


# Station belongs to one line
normal_stations = [
    station
    for station in stations
    if len(station_to_lines[station]) == 1
]

# Station belongs to two or more lines
interchange_stations = [
    station
    for station in stations
    if len(station_to_lines[station]) > 1
]


# Create Figure

fig, ax = plt.subplots(figsize=(18, 12))


# Draw Railway Lines

# These edges belong to both Bakerloo and Northern.
shared_bakerloo_northern_edges = [
    ("Charing Cross", "Embankment"),
    ("Embankment", "Waterloo")
]

for line_name, line_information in lines.items():

    normal_line_edges = [
        edge
        for edge in line_information["edges"]
        if edge not in shared_bakerloo_northern_edges
        and (edge[1], edge[0]) not in shared_bakerloo_northern_edges
    ]

    # Draw edges that are not shared
    nx.draw_networkx_edges(
        G,
        pos,
        edgelist=normal_line_edges,
        edge_color=line_information["color"],
        width=3,
        ax=ax
    )


# Draw shared Bakerloo and Northern lines as straight,
# parallel lines

shared_bakerloo_northern_edges = [
    ("Charing Cross", "Embankment"),
    ("Embankment", "Waterloo")
]


def draw_straight_parallel_line(
    ax,
    pos,
    station1,
    station2,
    color,  
    offset,
    width=3
):
    x1, y1 = pos[station1]
    x2, y2 = pos[station2]

    # Direction of the railway connection
    dx = x2 - x1
    dy = y2 - y1

    # Length of the connection
    length = (dx ** 2 + dy ** 2) ** 0.5

    # Perpendicular direction used to separate the two lines
    perpendicular_x = -dy / length
    perpendicular_y = dx / length

    # Apply the parallel offset
    shifted_x1 = x1 + perpendicular_x * offset
    shifted_y1 = y1 + perpendicular_y * offset

    shifted_x2 = x2 + perpendicular_x * offset
    shifted_y2 = y2 + perpendicular_y * offset

    ax.plot(
        [shifted_x1, shifted_x2],
        [shifted_y1, shifted_y2],
        color=color,
        linewidth=width,
        solid_capstyle="round",
        zorder=2
    )


for station1, station2 in shared_bakerloo_northern_edges:

    # Bakerloo Line - brown
    draw_straight_parallel_line(
        ax=ax,
        pos=pos,
        station1=station1,
        station2=station2,
        color=lines["Bakerloo"]["color"],
        offset=0.055
    )

    # Northern Line - black
    draw_straight_parallel_line(
        ax=ax,
        pos=pos,
        station1=station1,
        station2=station2,
        color=lines["Northern"]["color"],
        offset=-0.055
    )


# Draw Normal Stations
# All normal stations use circular nodes

for line_name, line_information in lines.items():

    stations_on_current_line = [
        station
        for station in normal_stations
        if station_to_lines[station][0] == line_name
    ]

    nx.draw_networkx_nodes(
        G,
        pos,
        nodelist=stations_on_current_line,
        node_color=line_information["color"],
        node_shape="o",             # Tất cả node là hình tròn
        node_size=450,
        edgecolors="white",
        linewidths=2,
        ax=ax
    )


# Draw Interchange Stations with Multiple Colours

interchange_radius = 0.20

for station in interchange_stations:

    x, y = pos[station]
    station_lines = station_to_lines[station]

    number_of_lines = len(station_lines)
    angle_per_line = 360 / number_of_lines

    # Each line receives one coloured section
    for index, line_name in enumerate(station_lines):

        start_angle = index * angle_per_line
        end_angle = (index + 1) * angle_per_line

        section = Wedge(
            center=(x, y),
            r=interchange_radius,
            theta1=start_angle,
            theta2=end_angle,
            facecolor=lines[line_name]["color"],
            edgecolor="none",
            zorder=6
        )

        ax.add_patch(section)

    # Outer border
    outer_circle = Circle(
        (x, y),
        radius=interchange_radius,
        facecolor="none",
        edgecolor="black",
        linewidth=2.2,
        zorder=7
    )

    ax.add_patch(outer_circle)

    # White central area
    inner_circle = Circle(
        (x, y),
        radius=interchange_radius * 0.38,
        facecolor="white",
        edgecolor="black",
        linewidth=1,
        zorder=8
    )

    ax.add_patch(inner_circle)


# Distance Labels

# Store one distance for each unique pair of stations
distance_by_pair = {}

for station1, station2, distance in edges:

    pair = tuple(sorted((station1, station2)))

    # Keep the first distance if the same connection
    # appears on more than one Underground line
    if pair not in distance_by_pair:
        distance_by_pair[pair] = distance


# These two connections share the Bakerloo and Northern lines.
# Their labels will be drawn manually.
shared_pairs = {
    tuple(sorted(("Charing Cross", "Embankment"))),
    tuple(sorted(("Embankment", "Waterloo")))
}


# Create automatic labels for all non-shared connections
edge_labels = {}

for station1, station2 in G.edges():

    pair = tuple(sorted((station1, station2)))

    # Skip the two shared connections
    if pair in shared_pairs:
        continue

    distance = distance_by_pair[pair]

    edge_labels[(station1, station2)] = f"{distance} km"


# Draw normal distance labels
nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=edge_labels,
    label_pos=0.5,
    font_size=8,
    font_color="#333333",
    rotate=False,
    bbox={
        "facecolor": "white",
        "edgecolor": "none",
        "alpha": 0.82,
        "pad": 1.5
    },
    ax=ax
)


# Function for placing a distance label above a line

def draw_distance_above_line(
    station1,
    station2,
    text,
    position=0.5,
    offset=0.22
):
    x1, y1 = pos[station1]
    x2, y2 = pos[station2]

    # Position along the railway connection
    label_x = x1 + (x2 - x1) * position
    label_y = y1 + (y2 - y1) * position

    # Direction of the connection
    dx = x2 - x1
    dy = y2 - y1

    line_length = (dx ** 2 + dy ** 2) ** 0.5

    # Perpendicular direction
    perpendicular_x = -dy / line_length
    perpendicular_y = dx / line_length

    # Move the label above the railway connection
    label_x += perpendicular_x * offset
    label_y += perpendicular_y * offset

    ax.text(
        label_x,
        label_y,
        text,
        fontsize=8,
        color="#333333",
        horizontalalignment="center",
        verticalalignment="center",
        bbox={
            "facecolor": "white",
            "edgecolor": "none",
            "alpha": 0.85,
            "pad": 1.5
        },
        zorder=15
    )


# Draw the two shared-route distance labels manually

draw_distance_above_line(
    station1="Charing Cross",
    station2="Embankment",
    text="0.4 km",
    position=0.50,
    offset=0.05
)

draw_distance_above_line(
    station1="Embankment",
    station2="Waterloo",
    text="0.7 km",
    position=0.50,
    offset=0.05
)

# Station Labels

for station, (label_x, label_y) in label_pos.items():

    ax.text(
        label_x,
        label_y,
        station,
        fontsize=9,
        fontweight=(
            "bold"
            if station in interchange_stations
            else "normal"
        ),
        horizontalalignment="center",
        verticalalignment="center",
        zorder=12
    )


# # Interchange Notes

# interchange_note_offsets = {
#     "Oxford Circus": (0.75, -0.55),
#     "Piccadilly Circus": (-1.90, -0.75),
#     "Tottenham Court Road": (-2.55, 0.90),
#     "Leicester Square": (0.45, -0.95),
#     "Charing Cross": (-1.95, -0.55),
#     "Embankment": (0.55, -0.40),
#     "Waterloo": (0.60, -1.00),
#     "Holborn": (0.45, 0.85)
# }


# Legend

legend_items = []

for line_name, line_information in lines.items():

    legend_items.append(
        Line2D(
            [0],
            [0],
            color=line_information["color"],
            linewidth=4,
            marker="o",                  # Hình tròn trong Key
            markerfacecolor=line_information["color"],
            markeredgecolor="white",
            markeredgewidth=1.5,
            markersize=10,
            label=line_name
        )
    )

legend_items.append(
    Line2D(
        [0],
        [0],
        marker="o",
        linestyle="None",
        markerfacecolor="white",
        markeredgecolor="black",
        markeredgewidth=2,
        markersize=11,
        label="Interchange Station (Multiple Lines)"
    )
)

ax.legend(
    handles=legend_items,
    title="Key",
    loc="lower right",
    frameon=True,
    fontsize=10,
    title_fontsize=11
)


# Final Display

ax.set_title(
    "London Underground Network",
    fontsize=18,
    fontweight="bold",
    pad=20
)

ax.set_aspect("equal")
ax.axis("off")

# Ensure stations at the edges are not cut off
ax.set_xlim(-1.3, 15.2)
ax.set_ylim(-1.2, 12.2)

plt.tight_layout()
plt.show()
```
